```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("../data/messy_sales_data.csv")

# Display first five records
display(df.head())
```

```python
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate records:", df.duplicated().sum())
```

```python
# Check missing values before cleaning
print("Missing values before cleaning:")
print(df.isnull().sum())

# Fill missing email values
df["Email"] = df["Email"].fillna("Not Provided")

# Fill missing city values
df["City"] = df["City"].fillna("Unknown")

# Fill missing quantity with median
df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)

# Fill missing unit price with median
df["Unit_Price"] = df["Unit_Price"].fillna(
    df["Unit_Price"].median()
)

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
df = pd.read_csv("../data/messy_sales_data.csv", sep="\t")

print("Columns:", df.columns.tolist())
display(df.head())
```

```python
print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate records:", df.duplicated().sum())
```

```python
# Fill missing values
df["Email"] = df["Email"].fillna("Not Provided")
df["City"] = df["City"].fillna("Unknown")

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].median())

# Remove duplicate records
df = df.drop_duplicates()

# Reset row numbers
df = df.reset_index(drop=True)

print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate records after cleaning:", df.duplicated().sum())

display(df.head())
```

```python
# Remove extra spaces and standardize text
df["Customer_Name"] = df["Customer_Name"].str.strip().str.title()
df["City"] = df["City"].str.strip().str.title()
df["Category"] = df["Category"].str.strip().str.title()
df["Email"] = df["Email"].str.strip().str.lower()

# Convert Order_Date into date format
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    format="%d-%m-%Y"
)

# Display standardized data
display(df.head(10))

print("Data standardization completed!")
```

```python
# Calculate total sales for each order
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

# Display the updated dataset
display(df.head(10))

# Calculate overall sales
total_revenue = df["Total_Sales"].sum()

print("Total Revenue:", round(total_revenue, 2))
```

```python
# Save the cleaned dataset
df.to_csv(
    "../cleaned_data/cleaned_sales_data.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
print("File location: cleaned_data/cleaned_sales_data.csv")
```

```python
# Generate sales summary report

total_orders = len(df)
total_revenue = df["Total_Sales"].sum()
average_order_value = df["Total_Sales"].mean()
total_quantity = df["Quantity"].sum()

print("========== SALES SUMMARY REPORT ==========")
print("Total Orders:", total_orders)
print("Total Revenue:", round(total_revenue, 2))
print("Average Order Value:", round(average_order_value, 2))
print("Total Quantity Sold:", total_quantity)
print("==========================================")

# Sales by category
category_report = df.groupby("Category")["Total_Sales"].sum().reset_index()

print("\nSales by Category:")
display(category_report)

# Sales by city
city_report = df.groupby("City")["Total_Sales"].sum().reset_index()

print("\nSales by City:")
display(city_report)
```

```python
import os

# Create the required folders if they do not exist
os.makedirs("../reports", exist_ok=True)
os.makedirs("../cleaned_data", exist_ok=True)

print("Project folders are ready!")
```

```python
# Save sales summary report

summary = pd.DataFrame({
    "Metric": [
        "Total Orders",
        "Total Revenue",
        "Average Order Value",
        "Total Quantity Sold"
    ],
    "Value": [
        total_orders,
        round(total_revenue, 2),
        round(average_order_value, 2),
        total_quantity
    ]
})

# Save reports as CSV files
summary.to_csv("../reports/sales_summary_report.csv", index=False)

category_report.to_csv(
    "../reports/category_sales_report.csv", index=False
)

city_report.to_csv(
    "../reports/city_sales_report.csv", index=False
)

print("All reports saved successfully!")
```

```python
import matplotlib.pyplot as plt
import os

# Make sure the reports folder exists
os.makedirs("../reports", exist_ok=True)

# Chart 1: Sales by Category
plt.figure(figsize=(8, 5))
plt.bar(category_report["Category"], category_report["Total_Sales"])
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("../reports/sales_by_category.png")
plt.show()
```

```python
# Chart 2: Sales by City
plt.figure(figsize=(8, 5))
plt.bar(city_report["City"], city_report["Total_Sales"])
plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("../reports/sales_by_city.png")
plt.show()
```

```python
# Chart 3: Daily Sales Trend
daily_sales = df.groupby("Order_Date")["Total_Sales"].sum().reset_index()

plt.figure(figsize=(10, 5))
plt.plot(daily_sales["Order_Date"], daily_sales["Total_Sales"], marker="o")
plt.title("Daily Sales Trend")
plt.xlabel("Order Date")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../reports/daily_sales_trend.png")
plt.show()
```

```python
# Create an Excel workbook with multiple sheets

with pd.ExcelWriter("../reports/automated_sales_report.xlsx") as writer:
    df.to_excel(writer, sheet_name="Cleaned Data", index=False)
    summary.to_excel(writer, sheet_name="Sales Summary", index=False)
    category_report.to_excel(writer, sheet_name="Category Sales", index=False)
    city_report.to_excel(writer, sheet_name="City Sales", index=False)

print("Excel report created successfully!")
```
